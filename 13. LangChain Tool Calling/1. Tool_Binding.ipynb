# **Tool Binding**

**Tool Binding** is the step where you register tools with a LLM so that:
1. The LLM knows what tools are available
2. It knows what each tool does ( via description )
3. It knows what input format to use( via schema )

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
from langchain.tools import tool
from dotenv import load_dotenv
import os
```

```python
# tool 

@tool
def multiply(a: int, b: int) -> int:
    "Given two numbers, return their product."
    return a * b
```

```python
result = multiply.invoke({"a": 6, "b": 7})

print(f"The result of multiplication is: {result}")
```

```python
print( multiply.name )
print( multiply.description )
print( multiply.args )
```

```python
load_dotenv()

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
llm_with_tools = llm.bind_tools([multiply])
```
