# **Tool Calling**

**Tool Calling** is the process where the LLM decides, during conversation or task that it needs to use a specific tool/function -- and generate a structued output with:

- the name of the tool
- and the arguments to call it with

The LLM does not actually run the tool -- it just suggests the tool and the input arguments. The actual execution is handled by Langchain or you.

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
from langchain.tools import tool
from dotenv import load_dotenv
import os
```

```python
# tool 

@tool
def multiply(a: int, b: int) -> int:
    "Given two numbers, return their product."
    return a * b
```

```python
load_dotenv()

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
llm_with_tools = llm.bind_tools([multiply])
```

```python
result = llm_with_tools.invoke('Hi, how are you?')

print(result)
```

```python
result = llm_with_tools.invoke('What is 12 multiplied by 13?')
print(result)
```

# **Tool Execution**

**Tool Execution** is the step where the actual Python function (tool) is run using the input arguments that the LLM suggested during tool calling.

In Simple words:

The LLM says : "Hey, call the **multiply** tool with a=8 and b=3."


**Tool Execution** is when you or Langchain actually run:
**multiply(a=8, b=3)**
and get the result 24.

```python
result.tool_calls[0]
```

```python
result.tool_calls[0]['args']
```

```python
multiply.invoke(result.tool_calls[0]['args'])
```

```python
multiply.invoke(result.tool_calls[0])
```
