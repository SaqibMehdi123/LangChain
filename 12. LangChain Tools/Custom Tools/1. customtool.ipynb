# **tool using @tool decorator**

```python
from langchain_core.tools import tool
```

```python
# step 1 - create a function
def multiply(a, b):
    """Multiply two numbers"""
    return a * b
```

```python
# step 2 - add type hints
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
# step 3 - add the tool decorator

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
result = multiply.invoke({"a": 5, "b": 7})
```

```python
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```
